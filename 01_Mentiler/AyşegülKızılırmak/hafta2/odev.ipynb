```python
def not_degerlendir(notu):
    if notu >= 90:
        return "AA"
    elif notu >= 70:
        return "BB"
    elif notu >= 50:
        return "CC"
    else:
        return "FF"

sonuc = not_degerlendir(75)
print(sonuc)
```

# Try-Except ile Hata Yönetimi

```python
try:
    sayi = int(input("Bir sayı girin: "))
    print("Girdiğiniz sayı:", sayi)
except:
    print("Lütfen sadece sayı girin.")
```

```python
try:
    sayi = int(input("Bir sayı girin: "))

    if sayi > 0:
        print("Sayı pozitiftir.")
    elif sayi < 0:
        print("Sayı negatiftir.")
    else:
        print("Sayı sıfırdır.")

except ValueError:
    print("Lütfen geçerli bir sayı girin.")
    
```

#  Koşullu Filtreleme Uygulaması

```python
sayilar = [12, 25, 8, 34, 17, 40, 5, 28]
```

```python
for sayi in sayilar:
    if sayi >= 20:
        print(sayi)
```

```python
def sayilari_filtrele(sayilar):
    for sayi in sayilar:
        if sayi >= 20:
            print(sayi)

sayilari_filtrele(sayilar)
```

```python
try:
    sinir = int(input("Kaçtan büyük sayıları görmek istiyorsunuz? "))

    for sayi in sayilar:
        if sayi >= sinir:
            print(sayi)

except ValueError:
    print("Lütfen bir sayı girin.")
```

```python
while True:
    try:
        sinir = int(input("Kaçtan büyük sayıları görmek istiyorsunuz? "))

        for sayi in sayilar:
            if sayi >= sinir:
                print(sayi)

        break

    except ValueError:
        print("Lütfen bir sayı girin.")
```

```python
def filtrele(sayilar, sinir):
    for sayi in sayilar:
        if sayi >= sinir:
            print(sayi)
```

```python
filtrele(sayilar, 20)
```
