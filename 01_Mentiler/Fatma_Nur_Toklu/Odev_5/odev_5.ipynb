```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
dataFrame = pd.read_csv("student_performance_dataset.csv")
```

```python
dataFrame
```

```python
cleanDataFrame = dataFrame.dropna()
```

```python
cleanDataFrame
```

```python
(figure, axes) = plt.subplots(nrows=1, ncols=3, figsize=(18, 5))

axes[0].hist(cleanDataFrame["final_exam_score"], bins=15, color="skyblue", edgecolor="black")
axes[0].set_title("Final Notlarının Dağılımı")
axes[0].set_xlabel("Final Sınav Notu")
axes[0].set_ylabel("Öğrenci Sayısı")

axes[1].scatter(cleanDataFrame["study_time_hours"], cleanDataFrame["final_exam_score"], color="purple", alpha=0.5)
axes[1].set_title("Çalışma Süresi ve Final Notu İlişkisi")
axes[1].set_xlabel("Haftalık Çalışma Saati")
axes[1].set_ylabel("Final Notu")

ilk_30_ogrenci = cleanDataFrame.head(30)

axes[2].plot(ilk_30_ogrenci["student_id"], ilk_30_ogrenci["previous_grade"], color="blue", marker="o", label="Önceki Not")
axes[2].plot(ilk_30_ogrenci["student_id"], ilk_30_ogrenci["final_exam_score"], color="red", marker="x", label="Final Notu")
axes[2].set_title("Öğrencilerin Önceki ve Final Notları")
axes[2].set_xlabel("Öğrenci ID")
axes[2].set_ylabel("Notlar")
axes[2].legend()
```
