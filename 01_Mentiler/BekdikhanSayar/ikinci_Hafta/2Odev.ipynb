```python
def Film_Izleme_Yası(yas):
    if yas < 0:
        return "HATA: Yaşınız sıfırdan küçük olamaz"
    elif yas >= 18:
        return "Yaşınız bu film için yeterli"
    else :
        return "Yaşınız bu filmi izlemek için çok küçük" 

while True:
    try:
        yas = int(input("Yaşınızı giriniz: (Programı durdurmak için 300 giriniz) "))
        if yas == 300:
            break
        else:
            print(Film_Izleme_Yası(yas))
    except:
        print("Yaşınızı düzgün giriniz")
```
