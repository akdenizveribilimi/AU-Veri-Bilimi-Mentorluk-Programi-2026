```python
import time
import numpy as np
```

```python
veri_miktari = 3000000
```

```python
maaslar_liste = list(range(20000, 20000 + veri_miktari))
maaslar_array = np.arange(20000, 20000 + veri_miktari)
```

## klasik yöntem

```python
baslangic = time.time()
```

```python
zamli_liste = []
for maas in maaslar_liste:
    zamli_liste.append(maas * 1.25)
```

```python
bitis = time.time()
python_suresi = bitis - baslangic
print(f"Klasik Döngü İle Hesaplama: {python_suresi:.4f} saniye")
```

## numpy ile 

```python
baslangic = time.time()
```

## karşılaştırma

```python
hiz_farki = python_suresi / numpy_suresi
print(f"\nÖzet: NumPy kütüphanesi işlemi {hiz_farki:.1f} kat daha kısa sürede tamamladı.")
```
